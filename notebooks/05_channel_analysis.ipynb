# Channel Performance Analysis

## Objective
Analyze marketing performance across different channels using key business metrics such as Revenue, Cost, Clicks, Leads, Conversions, and ROI.



```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/marketing_campaign_cleaned.csv")
```

```python
df.head()
```

```python
channel_summary = (
    df.groupby("Channel")
      .agg({
          "Revenue_USD": "sum",
          "Cost_USD": "sum",
          "Clicks": "sum",
          "Leads": "sum",
          "Conversions": "sum",
          "ROI": "mean"
      })
      .round(2)
)

channel_summary
```

```python
channel_summary.to_csv("../reports/channel_performance_summary.csv")

print("Channel Performance Summary saved successfully!")
```

```python
# Revenue by Channel

channel_summary["Revenue_USD"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Revenue by Channel")
plt.xlabel("Channel")
plt.ylabel("Revenue (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Average ROI by Channel

channel_summary["ROI"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average ROI by Channel")
plt.xlabel("Channel")
plt.ylabel("Average ROI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
