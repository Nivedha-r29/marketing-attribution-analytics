```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/marketing_campaign_cleaned.csv")
```

```python
channel_revenue = df.groupby("Channel")["Revenue_USD"].sum().sort_values(ascending=False)

print(channel_revenue)

channel_revenue.plot(kind="bar")
plt.title("Revenue by Marketing Channel")
plt.ylabel("Revenue")
plt.show()
```

```python
channel_roi = df.groupby("Channel")["ROI"].mean().sort_values(ascending=False)

print(channel_roi)

channel_roi.plot(kind="bar")
plt.title("Average ROI by Channel")
plt.ylabel("ROI")
plt.show()
```

```python
top_campaigns = df.sort_values(
    by="Revenue_USD",
    ascending=False
).head(10)

print(top_campaigns[
    ["CampaignID","Channel","Revenue_USD"]
])
```

```python
totals = [
    df["Impressions"].sum(),
    df["Clicks"].sum(),
    df["Leads"].sum(),
    df["Conversions"].sum()
]

stages = [
    "Impressions",
    "Clicks",
    "Leads",
    "Conversions"
]

plt.figure(figsize=(8,5))
plt.bar(stages, totals)
plt.title("Marketing Conversion Funnel")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Cost_USD",
    y="Revenue_USD",
    hue="Channel"
)

plt.title("Cost vs Revenue")
plt.show()
```

```python
conversion = df.groupby(
    "Channel"
)["Conversion_Rate"].mean()

print(conversion)

conversion.plot(kind="bar")
plt.title("Average Conversion Rate by Channel")
plt.show()
```
