```python
import pandas as pd

df = pd.read_csv("../data/raw/marketing_campaign.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values:\n")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df["StartDate"] = pd.to_datetime(df["StartDate"])
df["EndDate"] = pd.to_datetime(df["EndDate"])
```

```python
# Campaign Duration
df["Campaign_Duration"] = (
    df["EndDate"] - df["StartDate"]
).dt.days

# Cost Per Click
df["CPC"] = df["Cost_USD"] / df["Clicks"]

# Cost Per Lead
df["CPL"] = df["Cost_USD"] / df["Leads"]

# Conversion Rate
df["Conversion_Rate"] = (
    df["Conversions"] / df["Clicks"]
) * 100

df.head()
```

```python
df.to_csv(
    "../data/processed/marketing_campaign_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
